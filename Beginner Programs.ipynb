```python
# Greeting Program
name = str(input('Enter your name : '))
print(f'Welcome to {name}')
```

```python
# Maximum of two numbers
a = int(input('Enter first number : '))
b = int(input('Enter second number : '))

if(a>b):
    print(f'{a} is largest')

else:
    print(f'{b} is largest')
```

```python
# Even or Odd
num = int(input('Enter a number : '))

if(num % 2 == 0):
    print('Even')

else:
    print('Odd')
```

```python
# Positive, Negative or Zero
num = int(input('Enter a number : '))

if(num > 0):
    print('Positive')

elif(num < 0):
    print('Negative')

else:
    print('Zero')
```

```python
# Multiplication table
num = int(input('Enter a number : '))

for i in range(1,11):
    print(f'{num} x {i} = {num*i}')
```

```python
# Factorial 
num = int(input('Enter a number : '))

res = 1

while(num>1):
    res *= num
    num -= 1

print(f'Factorial is {res}')
```

```python
# Sum 1 to N
num = int(input('Enter a number : '))
res = 0

for i in range(1,num+1):
    res += i

print(res)
```

### Decision Statements

```python
# Simple If
name = 'Shreyas'

if(name == 'Shreyas'):
    print(True)
```

```python
# If-Else
name = 'Shreyas'

if(name == 'Shreyas'):
    print(True)

else:
    print(False)
```

```python
# If-Else-If
# Grade Program

marks = int(input('Enter marks of the student : '))

if(marks>90):
    print('A')

elif(marks>85):
    print('B')

elif(marks>70):
    print('C')

else:
    print('No Grade')
```
