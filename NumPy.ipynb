## Boolean Arrays

```python
x = np.random.randint(1,50,15)
x
```

```python
x.resize(3,5)
x
```

```python
barr = (x > 20)
barr
```

```python
x[barr]
```

```python
x[~barr]
```

```python
age = np.random.randint(1,21,25)
age.resize(5,5)
age
```

```python
ageb = (age > 18)
ageb
```

```python
age[ageb]
```

```python
age[~ageb]
```

```python
age[age > 18]
```

```python
zo = np.random.randint(0,2,25)
zo.resize(5,5)
zo
```

```python
zo[zo == 1].size
```

```python
zo[zo == 0].size
```

#### SYMBOLS -> &, |, ~

```python
sample = np.random.randint(0,3,30)
sample.resize(3,10)
sample
```

```python
sample[ (sample == 2)    |     (sample == 0) ]
```

```python
sample[     (sample > 0)           &         (sample < 2)      ]
```
