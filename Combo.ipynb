```python
#2 
list1 = [1,2,3,4,5]
list2 = [3,4,5,6,7]

def common(list1, list2):
    res = []

    for ele in list1:
        if(ele in list2):
            res.append(ele)

    return res

common(list1, list2)
```

```python
# 3 -> Student Management
# Exception Handling

sd = {}

def adds():
    global sd

    print('----------------------------------------')
    name = input('Enter Student\'s Name : ')
    marks = int(input('Enter Student\'s Marks : '))

    sd[name] = marks

    print(sd)
    print('--Student Added Successfully--')

def search():
    global sd

    print('----------------------------------------')
    name = input('Enter Student\'s Name to Search: ')

    try:
        print(f'{name} : {sd[name]}')
        

    except KeyError:
        print('--Student Record Doesn\'t Exist--')

def update():
    global sd

    print('----------------------------------------')
    name = input('Enter Student\'s Name to Update: ')
    marks = int(input('Enter New Marks : '))

    try:
        sd[name] = marks
        print('--Student Record is Updated--')

    except KeyError:
        print('--Student Record Doesn\'t Exist--')

def delete():
    global sd

    print('----------------------------------------')
    name = input('Enter Student\'s Name to Delete: ')

    try:
        del sd[name]
        print('--Student Record is Deleted--')

    except KeyError:
        print('--Student Record Doesn\'t Exist--')

def topper():
    global sd

    print('----------------------------------------')

    if(sd == {}):
        print('--Empty--')

    else:
        mlist = list(sd.values())
        mlist.sort()
        topmarks = mlist[-1]

        for key in sd:
            if(sd[key] == topmarks):
                print(f'Topper -> {key} : {sd[key]}') 
    
    
    
    
topper()
```

```python
# Student Management

while(True):
    print('----------------------------------------')

    print('-- STUDENT MANAGEMENT MENU --')
    print('1. Add a New Student')
    print('2. Update Student Marks')
    print('3. Delete Student Record')
    print('4. Find Highest Scorer')
    print('5. Search Student Record')
    print('0. STOP')

    o = int(input('Enter your choice : '))

    if(o == 0):
        break

    elif(o == 1):
        adds()

    elif(o == 2):
        update()

    elif(o == 3):
        delete()

    elif(o == 4):
        topper()

    elif(o == 5):
        search()

    else:
        print('--INVALID--')
    
```
