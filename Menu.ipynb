```python
# Simple Calculator

while(True):
    print('----------------------------------')
    a = int(input('Enter First Number : '))
    b = int(input('Enter Second Number : '))
    
    print('\n--MENU OPTIONS--')
    print('+ -> Addition')
    print('- -> Subtraction')
    print('* -> Multiplication')
    print('/ -> Division')
    
    o = input('\nEnter your choice (To stop enter BREAK) : ')
    res = None

    if(o == 'BREAK'):
        break
    
    elif(o == '+'):
        res = a+b
    
    elif(o == '-'):
        res = a-b
    
    elif(o == '*'):
        res = a*b
    
    elif(o == '/'):
        res = a/b

    else:
        res=None
    
    if(res == None):
        print('\n--ERROR--')
    
    else:
        print('Result : ',res)

    print('\n')
    

```

```python
# Stack Implementation

stack = list()

def sprint():
    global stack
    print('The Elements of Stack are : ', stack)

def push():
    global stack
    ele = int(input('Enter an element to push into stack : '))
    stack.append(ele)

def pop():
    global stack
    ele = stack[-1]
    stack = stack[:-1]
    print('The popped element is : ', ele)
```

```python
while(True):
    print('--------------------------------')

    print('--MENU--')
    print('1. Print the Stack')
    print('2. Push New Element')
    print('3. Pop An Element')
    print('0. Stop')

    o = int(input('\nEnter your option : '))

    if(o == 1):
        sprint()

    elif(o == 2):
        push()

    elif(o == 3):
        pop()

    elif(o == 0):
        break

    else:
        print('-- INVALID --')
    
```
