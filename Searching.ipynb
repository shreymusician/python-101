```python
# Linear Search / Sequential Search

n = int(input('Enter number of elements : '))
l = []

for i in range(0,n):
    ele = (input(f'Enter {i} index element : '))
    l.append(ele)

print('List : ', l)

key = (input('Enter key element : '))

for i in range(0,n):
    if(l[i] == key):
        print(f'{key} is matched in index {i}')
        break
```
