```python
# Leap Year
import calendar

year = 2024

if calendar.isleap(year):
    print('Leap')

else:
    print('Not Leap')
```

```python
# Palindrome
word = input('Enter a word : ')

l = list(word)
print(l)

rev = l[:]
rev.reverse()
print(rev)

if(l == rev):
    print('Palindrome')

else:
    print('Not Palindrome')
```

```python
# Given Number is Prime or Not
n = int(input('Enter a number : '))
flag = True # Given Number is Prime

if(n==1):
    print('Neither Prime Nor Composite')

else:   
    for i in range(2,n):
        if(n % i == 0):
            print('Not Prime')
            flag = False
            break

    if(flag):
        print('Prime')
```

```python
def isprime(n):
    if(n==1):
       return False

    else:   
        for i in range(2,n):
            if(n % i == 0):
                return False
        return True
```

```python
num = int(input('Enter a number : '))

for i in range(2,num+1):
    if(isprime(i)):
        print(i)
```

```python
# List Duplicates
l = [1,2,2,3,4]  # 5
# {1,2,3,4} - 4
def isunique(l): 
    s = set(l)

    if(len(l) == len(s)):
        return True

    else:
        return False

def numdup(l):
    s = set(l)
    return len(l)-len(s)

isunique(l)
numdup(l)
```

```python
# Count each element in a list using dictionary
a = [1,1,2,3,3,3,4,4]

def lcount(l):
    d = {}
    s = set(l)
    print(s)

    for ele in s:
        d[ele] = 0
    print(d)
    
    for key in d:
        for ele in a:
            if(ele == key):
                d[key]+=1
    print(d)

lcount(a)
```

```python
def iseven(num):
    if(num % 2 == 0):
        return True
    else:
        return False

def sq(num):
    return num**2
```

```python
l = [2,3,4,6,5]

def newlist(l):
    res = []

    for ele in l:
        if(iseven(ele)):
            res.append(ele)

    for i in range(0,len(res)):
        res[i] = sq(res[i])
    
    print(res)

newlist(l)
```
