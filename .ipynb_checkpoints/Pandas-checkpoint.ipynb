```python
import pandas as pd
```

#### Series, DataFrame, Panel
1. Series -> 1D ----- Column of Data
2. DataFrame -> 2D -------- Excel Sheet
3. Panel -> 3D ----------- Collection of Excel Sheet

```python
# Series
l = [2,3,4,2,3,4]
a = pd.Series(l)
a
```

```python
# Custom Indexing
b = pd.Series(['Shreyas', 'Druthi', 'Ambika'])
b
```

```python
b[0]
```

```python
b[1]
```

```python
c = pd.Series(['Shreyas', 'Druthi', 'Ambika'], index=['Python', 'Tally', 'Typing'])
c
```

```python
c['Python']
```

```python
c.index
```

```python
# Dict Example
d = {'Python':'Shreyas', 'Tally':'Druthi', 'Typing':'Ambika'}
d
```

```python
d.keys()
```

```python
d.values()
```

```python
ed = {'Apples' : 'Red', 'Bananas' : 'Yellow'}
ed.keys() # Index
```

```python
ed.values() # Column
```

```python
fruits = pd.Series(ed)
fruits
```

```python
fruits['Apples']
```

```python
fruits['Grapes'] = 'Green'
fruits
```

```python
del fruits['Apples']
fruits
```

```python
se = pd.Series([2,4,6,8,10], ['a','b','c','d','e'])
se
```

```python
se['a':'c']
```
