```python
# 2. No Input, Return

def sum2():
    a = 1
    b = 2
    return(a+b)

sum2()
```

```python
# 3. Input, No Return

def add(a,b):
    print(f'Sum of {a} and {b} is {a+b}') # Input - 2 numbers, Output - Print
```

```python
add(1,2)
```

```python
# 4. Input, Return

def add(a,b):
    return(a+b)
```

```python
add(1,2)
```

```python
# 5. Recursive Function
# Factorial - Loop, Recursive

def fact(n):
    if(n==1):
        return 1
        
    return n*fact(n-1)

fact(5)
```

```python
# Fibonacci

def fib(n):
    if(n==0):
        return 0

    elif(n==1):
        return 1

    else:
        return fib(n-2)+fib(n-1)

for i in range(0,10):
    print(fib(i))
```

```python
# 6. Lambda Functions

def sum(a,b):
    return a+b

sum(1,2)
```

```python
a = 1
b = 2

res = lambda a,b : a+b
res(a,b)
```

```python
res = lambda x : x

print(res('Hello World'))
```
