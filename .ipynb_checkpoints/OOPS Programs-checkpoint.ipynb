```python
# Shapes

class Shape:
    def SaboutMe(self):
        print('I am a Shape')

class Rectangle(Shape):
    l = None
    b = None
    
    def RaboutMe(self):
        print('I am a Rectangle')

    def __init__(self, length, breadth):
        self.l = length
        self.b = breadth

    def area(self):
        print('Area is : ', self.l * self.b)

    def per(self):
        print('Perimeter is : ', 2*(self.l + self.b))

class Square(Rectangle):
    n = None

    def SQaboutMe(self):
        print('I am a Square')
    
    def __init__(self, side):
        self.n = side

    def area(self):
        print('Area is : ', self.n**2)

    def per(self):
        print('Perimeter is : ', 4*self.n)

```

```python
sh = Shape()
```

```python
sh.SaboutMe()
```

```python
rec = Rectangle(10,5)
```

```python
rec.RaboutMe()
```

```python
rec.SaboutMe()
```

```python
rec.SQaboutMe()
```

```python
rec.area()
```

```python
rec.per()
```

```python
sq = Square(5)
```

```python
sq.SQaboutMe()
```

```python
sq.RaboutMe()
```

```python
sq.SaboutMe()
```

```python
sq.area()
```

```python
sq.per()
```

## Library System

```python
class Book(Library):
    title = None
    author = None

    isBorrowed = False

    def __init__(self):
        print('----------------------------------')
        self.title = input('Enter Book Title : ')
        self.author = input('Enter Book Author : ')        
        print('--Book Added Successfully--')

    def bookprint(self):
        print('----------------------------------')
        print(f'Title : {self.title}, Author : {self.author}')    
```

```python
def library():
    print('----------------------------------')
    nbooks = int(input('Enter Number of Books : '))

    bobj = []

    for i in range(nbooks):
        b = Book()
        bobj.append(b)

    print('\n')
    
    for i in range(nbooks):
        bobj[i].bookprint()

library()
```

## Nice Computers Program

```python
class Nice:
    cd = {'a' : 'Typing', 'b' : 'Computer Basics', 'c' : 'Tally', 'd' : 'Python'}
    ad = {'Typing' : 1000, 'Computer Basics' : 2000, 'Tally' : 2500, 'Python' : 2500}
    td = {'Typing' : 'Prakash Sir', 'Computer Basics' : 'Ambika Mam', 'Tally' : 'Ambika Mam', 'Python' : 'Shreyas Sir'}
    
    def cdisplay(self):
        print('----------------------------------------')
        print('WELCOME TO NICE COMPUTERS')
        print('--COURSE LIST--')

        print('a) Typing')
        print('b) Computer Basics')
        print('c) Tally')
        print('d) Python')
        

class Student(Nice):
    sname = None
    sclass = None
    choice  = None
    course = None
    

    def sdetails(self):
        print('----------------------------------------')
        self.sname = input('Enter your Name : ')
        self.sclass = input('Enter your Class : ')
        self.choice = input('Enter your choice : ')
        self.course = self.cd[self.choice]

    def adetails(self):
        print('----------------------------------------')
        print(f'Amount for the course {self.course} : {self.ad[self.course]}')

    def tdetails(self):
        print('----------------------------------------')
        print(f'Teacher Alloted for the course {self.course} : {self.td[self.course]}')

    def display(self):
        print('Student Name : ', self.sname)
        print('Class : ',self.sclass)
        print('Course : ',self.course)
    
    
```

```python
st = Student()
```

```python
st.cdisplay()
```

```python
st.sdetails()
```

```python
st.display()
```

```python
st.adetails()
```

```python
st.tdetails()
```
